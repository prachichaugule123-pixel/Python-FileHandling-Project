```python
#Q1. Create a student record system that allows adding student data (name, roll, marks) and saving the data to a file. 
# ***perfect*** 
#Tip:- (For Q5 multiple records are needed,  to to take multiple recordes run Q1 multiple times (minimum 3 times for 3 records)...

file = open("student.txt", "a")

name = input("Enter Name: ")
roll = input("Enter Roll No: ")
marks = input("Enter Marks: ")

file.write(name + ":" + roll + ":" + marks + "\n")

file.close()

print("Student Record Saved Successfully.")
```

```python
#Q2. Implement search functionality to find a student by roll number.
# ***perfect***


roll = input("Enter Roll No to Search: ")

file = open("student.txt", "r")

found = False

for line in file:
    data = line.strip().split(":")

    if len(data) == 3:
        if data[1] == roll:
            print("Name :", data[0])
            print("Roll :", data[1])
            print("Marks :", data[2])
            found = True
            break

if found == False:
    print("Student Not Found.")

file.close()
```

```python
#Q3. Add functionality to update student marks.
# ***perfect***


roll = input("Enter Roll No: ")
records = []
found = False
file = open("student.txt", "r")

for line in file:
    data = line.strip().split(":")

    if len(data) == 3:
        if data[1] == roll:
            found = True
            new_marks = input("Enter New Marks: ")
            data[2] = new_marks

        records.append(":".join(data))

file.close()

if found:
    file = open("student.txt", "w")

    for record in records:
        file.write(record + "\n")

    file.close()

    print("Marks Updated Successfully.")
else:
    print("Student Not Found.")
```

```python
#Q4. Delete a student record by name or roll number.
# ***perfect***

name = input("Enter Name to Delete: ")

records = []
found = False

file = open("student.txt", "r")

for line in file:
    data = line.strip().split(":")

    if len(data) == 3:
        if data[0] == name:
            found = True
        else:
            records.append(":".join(data))

file.close()

if found:
    file = open("student.txt", "w")

    for record in records:
        file.write(record + "\n")

    file.close()

    print("Student Record Deleted Successfully.")
else:
    print("Student Details Not Found.")
```

```python
#Q5. Show all student records sorted by marks.
# ***perfect***

students = []

file = open("student.txt", "r")

for line in file:
    data = line.strip().split(":")

    if len(data) == 3:
        students.append(data)

file.close()

students.sort(key=lambda x: int(x[2]), reverse=True)

print("Students Sorted By Marks")

for s in students:
    print(s[0], s[1], s[2])
```

```python
#Q6. Display student records with marks greater than 75.
# ***perfect***

file = open("student.txt", "r")

print("Students Scoring More Than 75")

for line in file:
    data = line.strip().split(":")

    if len(data) == 3:
        if int(data[2]) > 75:
            print(data[0], data[1], data[2])

file.close()
```

```python
#Q7. Export top 3 performing students to a new file.
# ***perfect***

students = []

file = open("student.txt", "r")

for line in file:
    data = line.strip().split(":")

    if len(data) == 3:
        students.append(data)

file.close()

students.sort(key=lambda x: int(x[2]), reverse=True)

file = open("top3.txt", "w")

for student in students[:3]:
    file.write(":".join(student) + "\n")

file.close()

print("Top 3 Students Exported.")
```

```python
#Q9. Count how many students scored more than 90 and less than 35.
# ***perfect***

above90 = 0
below35 = 0

file = open("student.txt", "r")

for line in file:
    data = line.strip().split(":")

    if len(data) == 3:
        marks = int(data[2])

        if marks > 90:
            above90 += 1

        if marks < 35:
            below35 += 1

file.close()

print("Students Scored Above 90 =", above90)
print("Students Scored Below 35 =", below35)
```
